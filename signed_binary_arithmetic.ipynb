```python
# Question 2: Rotate Operations on 8-bit Binary Numbers

def rotate_left_without_carry(binary, positions):
    print("\nRotate Left Without Carry (ROL)")
    print("Original number:", binary)

    for i in range(positions):
        first_bit = binary[0]
        binary = binary[1:] + first_bit
        print("Step", i + 1, ":", binary)

    print("Final result:", binary)


def rotate_left_with_carry(binary, carry, positions):
    print("\nRotate Left With Carry (RCL)")
    print("Original number:", binary)
    print("Initial carry:", carry)

    for i in range(positions):
        first_bit = binary[0]
        binary = binary[1:] + str(carry)
        carry = int(first_bit)

        print("Step", i + 1, ":", binary,
              "| Carry =", carry)

    print("Final result:", binary)
    print("Final carry:", carry)


# (a) ROL 11110110 by 2 positions
rotate_left_without_carry("11110110", 2)

# (b) RCL 10000110 by 2 positions
# Initial carry is assumed to be 0
rotate_left_with_carry("10000110", 0, 2)
```
